```python
import numpy as np
import matplotlib.pyplot as plt
import thermodynamics as thermo
```

```python
T=np.logspace(-2,1,50)
y=np.zeros(len(T))
for i in range(len(T)):
    y[i]=thermo.p(T[i],0.511,2)/(T[i]**4)
plt.figure()
plt.semilogx(T,y)
plt.xlim(10,0.01)
plt.xlabel('T')
plt.ylabel('p/T^4')
```

```python
T=np.logspace(-2,1,50)
z=np.zeros(len(T))
for i in range(len(T)):
    z[i]=thermo.s(T[i],0.511,2)/(T[i]**3)
plt.figure()
plt.semilogx(T,z)
plt.xlim(10,0.01)
plt.xlabel('T')
plt.ylabel('s/T^3')
plt.show()
```
