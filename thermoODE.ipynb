```python
import numpy as np
import matplotlib.pyplot as plt
import thermodynamics as thermo
import ODESolve_slow as ODE

```

```python
def f(a,y,p):
   
    T, p, P, s = y  # y is an array of length 4
    
    der = np.zeros(4) # der is an array of length 4 
    der[0] = -T/a
    der[1] = thermo.dpdT(T,0.511,2)*der[0]
    der[2] = thermo.dPdT(T,0.511,2)*der[0]
    der[3] = thermo.dsdT(T,0.511,2)*der[0]
    
    return der

a0=0.1
y0=np.array([10,thermo.p(10,0.511,2),thermo.P(10,0.511,2),thermo.s(10,0.511,2)])
dx0=0.001
p=None


N_step = 1000
dN = 2
x_final = 100


print(f(a0,y0,p)) 

a, y, dx, done = ODE.ODEOneRun(a0, y0, dx0, p, N_step, dN, x_final)
print(done)
```

```python
plt.figure()
plt.semilogx(y[:,0],y[:,1])
plt.xlim(10,0.01)

T=np.logspace(-2,1,50)
z=np.zeros(len(T))
for i in range(len(T)):
    z[i]=thermo.p(T[i],0.511,2)
plt.semilogx(T,z,linestyle='--')
plt.xlim(10,0.01)
```
