```python
import numpy as np
import matplotlib.pyplot as plt
x, w = np.polynomial.laguerre.laggauss(50)
xj, wj = np.polynomial.legendre.leggauss(50)
from constants import gA,GF, dmnp, me, hbar
```

```python
#Gauss Legendre

def I1(A):
    
    g=np.zeros(50)

    for j in range(len(g)):
        E=(A/2)*(xj[j]+1)
        g[j]=np.exp(-E)
        
    return (A/2)*np.sum(wj*g)

print(I1(1))
1-np.exp(-1)
```

```python
#Gauss Laguerre

def I2(B):

    g=np.zeros(50)

    for j in range(len(g)):
        E=x[j]+B
        g[j]=np.exp(-E)

    return np.sum(w*g*np.exp(x))

print(I2(1))
np.exp(-1)

```

```python
#Gauss Laguerre Test 2

def I2(B):

    g=np.zeros(50)

    for j in range(len(g)):
        E=x[j]+B
        g[j]=E*np.exp(-E**2)

    return np.sum(w*g*np.exp(x))

print(I2(1))
0.5*np.exp(-1)
```

```python
#Free neutron decay rate using Gauss Legendre
def I1(A):
    
    g=np.zeros(50)

    for j in range(len(g)):
        E=(A/2)*(xj[j]+1)
        g[j]=E**2*(dmnp-E)*np.sqrt((dmnp-E)**2-me**2)
        
    return (A/2)*np.sum(wj*g)
rate=I1(dmnp-me)*(GF**2*(1+3*gA**2)/(2*np.pi**3))
```

```python

Hz=rate/hbar
```
